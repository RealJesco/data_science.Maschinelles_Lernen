Die automatisierte Klassifikation von Vogelgeräuschen hat sich in den letzten Jahren zu einem etablierten Forschungsfeld im Schnittbereich von Bioakustik, Mustererkennung und Deep Learning entwickelt. Im aktuellen Forschungsstand lassen sich drei zentrale Entwicklungslinien erkennen:
1.   die Weiterentwicklung von Modellarchitekturen,

Frühere Arbeiten zeigten bereits, dass Convolutional Neural Networks (CNN) klassische Verfahren wie SVM oder HMM in der Vogelstimmenklassifikation übertreffen (vgl. Hidayat et al., 2021; Carvalho & Gomes, 2022). CNNs eignen sich insbesondere deshalb, weil spektrale Darstellungen von Audiosignalen als zweidimensionale Zeit-Frequenz-Bilder interpretiert werden können.

Permana et al. (2021) setzen CNN-Modelle ein, um Spektrogramm-basierte Eingaben zu klassifizieren. Die Transformation der Rohdaten wird dabei als vorbereitender Standardprozess behandelt und nicht als eigenständige Untersuchungsvariable.

Pan et al. (2024) entwickeln ein MFCC-basiertes CNN-System zur Erkennung von Vogelrufen im Kontext der Stromleitungsüberwachung. Die Arbeit beschreibt detailliert die MFCC-Extraktion (Vorverarbeitung, DFT, Mel-Filterbank, DCT) und zeigt robuste Ergebnisse auch unter Rauschbedingungen. Allerdings wird ausschließlich MFCC als Feature genutzt; alternative Repräsentationen werden nicht systematisch verglichen.

Diese Studien verdeutlichen exemplarisch, dass CNN-basierte Modelle inzwischen als methodischer Standard gelten.

---

Ein zentrales Element nahezu aller aktuellen Ansätze ist die Transformation des Audiosignals in eine geeignete numerische oder visuelle Repräsentation vor der eigentlichen Modellierung. Roh-Audiosignale werden typischerweise in Zeit-Frequenz-Darstellungen überführt, um frequenzabhängige Muster sichtbar zu machen. Besonders häufig werden Mel-Spektrogramme und Mel Frequency Cepstral Coefficients (MFCC) eingesetzt (vgl. Carvalho & Gomes, 2022; Hidayat et al., 2021). Mel-Spektrogramme bilden die Frequenzachse auf einer psychoakustisch motivierten Mel-Skala ab und ermöglichen CNNs eine bildähnliche Verarbeitung. MFCC hingegen liefern eine kompaktere, niedrigdimensionale Repräsentation des Spektrums und gelten als recheneffizient sowie vergleichsweise robust gegenüber Störgeräuschen (vgl. Carvalho & Gomes, 2022).

Neben diesen etablierten Verfahren finden auch Per-Channel Energy Normalization (PCEN) sowie teilweise Raw-Waveform-Ansätze Anwendung (vgl. Dias et al., 2025; Heinrich et al., 2025). PCEN wird insbesondere als robuste Alternative zum Mel-Spektrogramm diskutiert, da es eine integrierte Normalisierung und Rauschunterdrückung beinhaltet (vgl. Dias et al., 2025). Andere Arbeiten kombinieren mehrere Repräsentationen oder ergänzen spektrale Darstellungen durch akustische Indizes oder handgefertigte Features (vgl. Dias et al., 2025). Auch klassische Feature-Engineering-Ansätze auf Basis von MFCC, Wavelet-Transformationen oder texturbasierten Merkmalen zeigen weiterhin hohe Klassifikationsleistungen, insbesondere in ressourcenbeschränkten Szenarien (vgl. Celik & Akbal, 2026; Carvalho & Gomes, 2022).

Die am häufigsten verwendeten Repräsentationen sind somit:
- Log-Mel-Spektrogramme
- MFCC
- Raw-Waveform
- PCEN

Ein wiederkehrendes Ergebnis der Literatur ist, dass die Wahl der Audio-Transformation maßgeblichen Einfluss auf die Klassifikationsleistung hat. Unterschiedliche Repräsentationen betonen jeweils verschiedene Eigenschaften des Signals:
- Spektrogramme heben zeitlich-dynamische Frequenzverläufe hervor,
- MFCC komprimieren spektrale Information in eine kompakte Form,
- PCEN integriert eine adaptive Energie-Normalisierung,
- Wavelet-basierte Verfahren erfassen mehrskalige Zeit-Frequenz-Strukturen.

---

Obwohl der aktuelle Forschungsstand die hohe Relevanz der Eingaberepräsentation wiederholt betont, bleibt deren isolierter Einfluss bislang nur unzureichend systematisch untersucht. Zahlreiche Arbeiten zeigen, dass spektrale Transformationen wie Mel-Spektrogramme oder MFCC leistungsfähige und praktikable Grundlagen für CNN-basierte Klassifikationssysteme darstellen (vgl. Hidayat et al., 2021; Carvalho & Gomes, 2022; Pan et al., 2024). Gleichzeitig weisen neuere Studien darauf hin, dass alternative oder optimierte Zeit-Frequenz-Repräsentationen, etwa PCEN oder spezifisch angepasste Frequenzskalierungen, signifikante Auswirkungen auf Genauigkeit und Robustheit haben können (vgl. Dias et al., 2025; Xie & Zhu, 2022). Damit besteht in der Literatur Einigkeit darüber, dass die Wahl der Transformation nicht lediglich ein vorbereitender Schritt, sondern ein zentraler Einflussfaktor für die Modellperformance ist.

Dennoch wird dieser Einfluss in den meisten Arbeiten nicht unter kontrollierten Bedingungen isoliert analysiert. Häufig werden mehrere Variablen gleichzeitig verändert, etwa Architektur, Datenaufbereitung, Trainingsstrategie oder Feature-Kombinationen (vgl. Dias et al., 2025; Xie & Zhu, 2022). Auch Studien, die unterschiedliche Repräsentationen einsetzen, koppeln diese meist an spezifische Netzwerkvarianten oder Datensätze, sodass keine direkte Vergleichbarkeit unter identischen Rahmenbedingungen gegeben ist (vgl. Pan et al., 2024; Permana et al., 2021). Dadurch bleibt offen, in welchem Ausmaß Leistungsunterschiede tatsächlich auf die gewählte Audio-Transformation zurückzuführen sind und nicht auf architekturelle oder trainingsbezogene Anpassungen.

Die bestehende Forschung liefert somit wertvolle Einzelbefunde zur Leistungsfähigkeit bestimmter Repräsentationen, jedoch fehlt eine systematische Gegenüberstellung zentraler Audio-Transformationen unter identischer Modellarchitektur und konsistenter Trainingskonfiguration.

##2.1 Beschreibung des HSN-Datensatzes

Für die vorliegende Untersuchung wird der HSN-Datensatz (High Sierra Nevada) verwendet, der von Clapp et al. (2023) auf Zenodo veröffentlicht wurde. Der Datensatz umfasst 100 vollständig annotierte Soundscape-Aufnahmen mit einer Länge von jeweils zehn Minuten, die im Jahr 2015 im südlichen Teil der Sierra Nevada in Kalifornien (USA) aufgezeichnet wurden. Insgesamt enthalten die Aufnahmen 10.296 manuell gesetzte Bounding-Box-Annotationen für 21 verschiedene Vogelarten aus dem Westen der Vereinigten Staaten. Der Datensatz wurde unter anderem als Testdatensatz im Rahmen der BirdCLEF- und Kaggle-Birdcall-Identifikation verwendet und eignet sich somit insbesondere für die Evaluation maschineller Lernverfahren.

Die Aufnahmen stammen aus hochgelegenen Regionen (über 3.000 m) in den Sequoia- und Kings-Canyon-Nationalparks und wurden mithilfe von Song-Meter-SM2+-Geräten in Mono mit einer ursprünglichen Abtastrate von 48 kHz aufgezeichnet. Für die veröffentlichte Version wurden die Daten auf 32 kHz resampelt und in das verlustfreie FLAC-Format konvertiert. Die Annotationen wurden mithilfe der Software Raven Pro erstellt und enthalten für jeden Vogelruf Zeit- und Frequenzbegrenzungen sowie einen eBird-Arten-Code gemäß der Clements-Taxonomie (2021). Neben den Audiodateien stehen strukturierte Metadaten in Form von CSV-Dateien zur Verfügung, die Informationen zu Zeitintervallen, Frequenzbereichen, Artenzuordnung sowie Aufnahmeort enthalten.

##2.2 Struktur der Rohdaten

Der HSN-Datensatz liegt in einer klar strukturierten Dateiorganisation vor, die eine getrennte Verarbeitung von Audiodaten, Annotationen und Metadaten ermöglicht. Die Audioaufnahmen sind als einzelne FLAC-Dateien abgelegt, wobei jede Datei eine eindeutige Kennzeichnung enthält, bestehend aus einer fortlaufenden ID sowie Datum und Uhrzeit der Aufnahme. Diese Namenskonvention erlaubt eine eindeutige Zuordnung zwischen Audiodateien und den entsprechenden Annotationseinträgen.

Die zugehörigen Annotationen sind in der Datei *annotations.csv* gespeichert. Jede Zeile beschreibt ein einzelnes annotiertes Schallereignis und enthält neben dem Dateinamen die Start- und Endzeit in Sekunden sowie die untere und obere Frequenzgrenze in Hertz. Zusätzlich wird jeder Eintrag durch einen eBird-Arten-Code klassifiziert, wodurch eine eindeutige Zuordnung zu einer Vogelart möglich ist. Nicht eindeutig identifizierbare Lautäußerungen sind gesondert gekennzeichnet und bleiben als eigenständige Kategorie im Datensatz erhalten.

Ergänzend dazu ermöglicht die Datei *species.csv* die Zuordnung der Arten-Codes zu wissenschaftlichen und gebräuchlichen Artnamen. Weitere Metainformationen, etwa zu den geografischen Koordinaten der Aufnahmeorte, sind in separaten Textdateien dokumentiert. Diese modulare Struktur bildet die Grundlage für die anschließende Segmentierung der Audiodaten sowie für die gezielte Generierung unterschiedlicher Audio-Transformationen.

### 2.3.2 Klassenverteilung

Zur Beurteilung der Datenstruktur wird zunächst die Verteilung der Annotationen über die einzelnen Vogelarten genauer untersucht. Eine unausgewogene Klassenverteilung, wie  kann die Modellperformance beeinflussen und ist daher für die spätere Trainingsstrategie relevant.

```python
# Grundlegende Kennzahlen zur Verteilung
min_count = class_counts.min()
max_count = class_counts.max()
ratio = max_count / min_count

print("Minimum (seltenste Klasse):", min_count)
print("Maximum (häufigste Klasse):", max_count)
print("Verhältnis max/min:", round(ratio, 2))

# Anteil der Top-5-Klassen an allen Annotationen
top5_share = class_counts.head(5).sum() / class_counts.sum()
print("Anteil der Top-5-Klassen:", round(top5_share * 100, 2), "%")
```

```python
# Berechnung der prozentualen Anteile pro Klasse
class_percent = (class_counts / class_counts.sum() * 100).round(2)

# Formatierung der Werte als Prozentangabe für die Ausgabe
class_percent = class_percent.astype(str) + " %"
class_percent
```

 Nachdem ein Schwellenwert gesetzt wurde und Klassen unterhalb diesem zu „Sonstiges“ zusammengefasst wurden, stehen mehrere zusätzliche Hebel zur Verfügung, um die verbleibende Unwucht im Training und in der Evaluation sinnvoll zu behandeln.  

 *Umgang mit Klassenungleichverteilung: Optionen nach dem Setzen eines Schwellenwerts*

1) Class Weights / gewichtete Loss

Eine naheliegende Strategie ist, Fehler bei seltenen Klassen stärker zu bestrafen als Fehler bei häufigen Klassen. Dazu werden Klassen-Gewichte in die Loss-Funktion eingebaut, bspw. invers proportional zur Häufigkeit oder nach einer geglätteten Variante wie „effective number of samples“.
Vorteilhaft ist dabei, dass das Modell gezielt dazu gedrängt wird, seltene Klassen nicht zu ignorieren, ohne dass wir Daten duplizieren müssen.
Zu aggressive Gewichte können das Training jedoch instabil machen oder die Performance auf häufigen Klassen deutlich verschlechtern. Oft ist daher eine moderate Gewichtung bzw. ein Tuning der Gewichte sinnvoll.

2) Balanced Sampling / balanced mini-batches

Statt die Trainingsdaten in ihrer natürlichen Verteilung zu sampeln, kann man das Sampling so anpassen, dass seltene Klassen häufiger in den Mini-Batches erscheinen, Oversampling, bzw. häufige Klassen etwas seltener, Undersampling. Eine verbreitete Praxis sind balanced mini-batches, in denen pro Batch ungefähr gleich viele Beispiele pro Klasse vorkommen.
Seltene Klassen werden dadurch im Training regelmäßig gesehen, was besonders bei sehr dominanten Klassen hilft.
Zu beachten ist, dass reines Oversampling Overfitting auf wenigen Beispielen fördern kann, da ähnliche/gleiche Samples oft wiederkehren. Funktioniert am besten in Kombination mit Augmentierung oder moderater Gewichtung.

3) Augmentierung (künstliche Variation vorhandener Beobachtungen)

Unter Augmentierung versteht man, dass man aus vorhandenen Trainingsbeobachtungen leicht veränderte Versionen erzeugt, um die effektive Datenmenge zu erhöhen und die Generalisierung zu verbessern. Je nach Datentyp kann das unterschiedlich aussehen, bspw. bei Audio: Zeitverschiebung, Noise-Mischung, SpecAugment; bei Bildern: Crop/Flip/Helligkeit.
Besonders bei seltenen Klassen kann Augmentierung helfen, dass das Modell robustere Merkmale lernt und nicht nur „auswendig“ lernt.
Schlechte oder zu starke Augmentierungen können das Signal jedoch verfälschen und Leistung verschlechtern. Wichtig ist außerdem: Augmentierung gehört nur ins Training, nicht in Validation/Test.

4) Optional zweistufig: „bekannt vs. unknown“, anschließend Artenklassifikation

Eine weitere Option ist ein zweistufiges Verfahren. In Stufe 1 entscheidet ein Modell oder ein Schwellenwert auf der Modellkonfidenz, ob ein Beispiel zu den "bekannten" Klassen gehört oder eher "Sonstiges" ist. Nur wenn „bekannt“ vorhergesagt wird, folgt Stufe 2: die eigentliche Spezies-Klassifikation innerhalb der bekannten Klassen.
Dies reduziert falsche Zuordnungen, und behandelt "Sonstiges" konzeptionell sauberer.
Jedoch wird eine zusätzliche Fehlerquelle eingebaut: Wenn Stufe 1 ein eigentlich bekanntes Beispiel als "Sonstiges" abweist, kann Stufe 2 es nicht mehr korrekt klassifizieren. Daher müssen Schwellenwerte und Trade-offs gut abgestimmt werden.

5) Passende Evaluation: Macro-F1 / Balanced Accuracy + per-class Recall statt nur Accuracy

Bei unausgeglichenen Daten ist Accuracy oft irreführend: Ein Modell kann sehr gut aussehen, obwohl es seltene Klassen fast nie erkennt. Deshalb sollten wir Metriken nutzen, die seltene Klassen sichtbar machen.
	•	Macro-F1 gewichtet alle Klassen gleich, unabhängig von Häufigkeit, und ist daher sehr geeignet, um „Long Tail“-Probleme aufzudecken.
	•	Balanced Accuracy, also der mittlere Recall über Klassen, ist ebenfalls robust gegenüber Ungleichverteilung.
	•	Per-class Recall/Precision und Confusion-Matrix helfen zusätzlich bei der Fehleranalyse. Es betrachtet welche Arten nie erkannt werden und welche verwechselt werden.
Ziel: Nicht nur „gute Gesamtzahl“, sondern ein realistisches Bild davon, ob seltene Arten überhaupt sinnvoll erkannt werden.

**Festlegung der Segmentlänge**

Im Rahmen der Datenvorverarbeitung werden die 10-minütigen FLAC-Soundscape-Aufnahmen in feste Einzel-Segmente überführt. Ziel ist es, für jede Annotation ein standardisiertes Audiosignal zu erzeugen, das als Eingabe für die anschließenden Audio-Transformationen und das CNN dient. Hierfür wird eine einheitliche Segmentlänge von **2 Sekunden** festgelegt.

Die Wahl dieser Segmentlänge basiert auf der zuvor durchgeführten Daueranalyse (Kapitel 2.3.3). Der Median der Segmentdauer liegt bei 0,3 Sekunden, 75 % aller Annotationen sind kürzer als 0,9 Sekunden und lediglich 6,32 % überschreiten eine Dauer von 2 Sekunden. Segmente mit mehr als 3 Sekunden machen nur 3,46 % aus, während sehr lange Annotationen (z. B. > 5 Sekunden) mit 1,83 % eine deutliche Minderheit darstellen. Die überwiegende Mehrheit der Vogelrufe ist somit sehr kurz.

Vor diesem Hintergrund stellt eine Segmentlänge von 2 Sekunden einen geeigneten Kompromiss dar, da einerseits über 93 % aller Annotationen vollständig innerhalb des Fensters abgebildet werden und andererseits vermieden wird, unnötig viel Hintergrundgeräusch oder zusätzliche akustische Ereignisse in das Segment aufzunehmen, was bei längeren Fenstern zu einer erhöhten Störanfälligkeit führen könnte.

Die Segmentierung erfolgt, indem für jede Annotation ein 2-Sekunden-Fenster um den Mittelpunkt der annotierten Zeitspanne zentriert wird. Dadurch wird sichergestellt, dass der akustische Kern des Rufes im Zentrum des Segments liegt. Segmente, die am Anfang oder Ende einer Aufnahme liegen und nicht vollständig innerhalb des definierten Fensters extrahiert werden können, werden mittels Zero-Padding auf die erforderliche Länge ergänzt. Auf diese Weise bleibt die Eingabedimension konstant, ohne die zeitliche Position des Rufes zu verschieben.

Annotationen mit einer Dauer von mehr als 2 Sekunden werden nicht ausgeschlossen, sondern entsprechend auf das definierte Fenster zugeschnitten. Da lediglich 6,32 % der Segmente diese Länge überschreiten, betrifft diese Kürzung nur einen kleinen Anteil des Datensatzes und beeinträchtigt die Gesamtstruktur nicht wesentlich. Insgesamt wird damit eine einheitliche, methodisch kontrollierte Datengrundlage geschaffen, die eine faire Vergleichbarkeit der unterschiedlichen Audio-Transformationen sowie eine stabile Modellarchitektur gewährleistet.

Bei sehr kurzen Vogelrufen, beispielsweise mit einer Dauer von 0,5 Sekunden, erfolgt keine zeitliche Streckung oder künstliche Verlängerung des Signals. Stattdessen wird das 2-Sekunden-Fenster um den Mittelpunkt der Annotation zentriert, sodass der Ruf im Zentrum des Segments liegt. Der verbleibende Teil des Fensters besteht aus dem originalen Umgebungssignal der jeweiligen Aufnahme. Eine Veränderung der Samplingrate oder der zeitlichen Struktur findet nicht statt. Auf diese Weise bleibt das akustische Signal unverfälscht, während gleichzeitig eine einheitliche Eingabedimension für das Modell gewährleistet wird und dieses den Call im Kontext lernen kann.

**Auswirkungen auf das Projekt**

Die Festlegung einer festen Segmentlänge hat mehrere zentrale Auswirkungen:
- Einheitliche Eingabedimension: Alle Audio-Transformationen basieren auf identisch langen Audiosignalen. Dadurch wird eine faire Vergleichbarkeit der Transformationen gewährleistet.
- Kontrollierte Modellarchitektur: Das CNN erhält konsistente Eingabegrößen, wodurch architekturelle Anpassungen vermieden werden können.
- Begrenzung von Hintergrundanteilen: Durch das relativ kurze Fenster wird der Anteil nicht relevanter Umgebungsgeräusche reduziert.
- Rechenleistung: Kürzere Segmente reduzieren Speicher- und Trainingsaufwand, was insbesondere bei über 10.000 Annotationen relevant ist.

```python
# Segmentlänge festlegen (2 Sekunden)
segment_length_sec = 2
```

```python
# Segmentierungsfunktion
def extract_segment(audio, sr, start, end):
    """
    Extrahiert ein zentriertes Segment fester Länge (2 Sekunden)
    um den Mittelpunkt einer Annotation.
    """

    segment_samples = int(segment_length_sec * sr)

    # Mittelpunkt berechnen
    mid = (start + end) / 2
    mid_sample = int(mid * sr)

    start_sample = mid_sample - segment_samples // 2
    end_sample = mid_sample + segment_samples // 2

    # Padding falls notwendig
    if start_sample < 0:
        segment = np.pad(audio[0:end_sample], (abs(start_sample), 0))
    elif end_sample > len(audio):
        segment = np.pad(audio[start_sample:len(audio)], (0, end_sample - len(audio)))
    else:
        segment = audio[start_sample:end_sample]

    return segment
```

```python
# Segmentierung durchführen
audio_path = "/content/data/audio"

X_segments = []
y_labels = []

# Gruppieren nach Dateiname
grouped_annotations = annotations.groupby("Filename")

for filename, group in tqdm(grouped_annotations):

    file_path = os.path.join(audio_path, filename)

    #sfo =soundfile.Soundfile(librosa.ex(file_path))

    # Audio-Datei einmal laden
    audio, sr = librosa.load(file_path, sr=None)
    #audio, sr = soundfile.read(file_path)

    for _, row in group.iterrows():
        segment = extract_segment(
            audio,
            sr,
            row["Start Time (s)"],
            row["End Time (s)"]
        )

        X_segments.append(segment)
        y_labels.append(row["Species eBird Code"])
```

```python
# In NumPy-Arrays umwandeln
X_segments = np.array(X_segments)
y_labels = np.array(y_labels)

print("Segment-Array Shape:", X_segments.shape)
print("Label-Array Shape:", y_labels.shape)
```

##3.3 Feature Generierung / Audio-Transformation

Damit die Auswahl der anschließend untersuchten drei Repräsentationen nachvollziehbar ist, werden zunächst vier gängige Verfahren skizziert und gegenübergestellt, bevor auf Basis ihrer Unterschiede die finale Auswahl getroffen wird.
Für die Klassifikation von Vogelstimmen lassen sich Audiodaten auf unterschiedliche Weise darstellen, wobei sich die Verfahren vor allem darin unterscheiden, wie stark sie das Rohsignal vorstrukturieren und wie robust sie gegenüber wechselnden Aufnahmebedingungen sind.

**Log-Mel-Spektrogramme** entstehen, indem das Signal in kurze Zeitfenster zerlegt und in ein Zeit-Frequenz-Bild überführt wird, die Frequenzachse anschließend auf die Mel-Skala abgebildet und die Energien logarithmisch komprimiert werden, wodurch typische Muster von Vogelrufen wie Harmonische, Triller und Anstiege sehr gut sichtbar werden und sich mit CNNs meist zuverlässig lernen lassen, während die Darstellung ohne zusätzliche Normalisierung dennoch empfindlich auf Hintergrundrauschen, unterschiedliche Recorder und stark schwankende Lautstärken reagieren kann.

**MFCC** gehen von einer ähnlichen Grundlage aus, da sie aus den log-skalierten Mel-Energien berechnet werden, diese jedoch weiter stark verdichten, indem sie über eine Kosinustransformation in wenige Koeffizienten zusammengefasst werden, was die Merkmalsdimension reduziert und Training oft stabil und schnell macht, zugleich aber feine spektrale Details glätten kann, die bei Vogelstimmen relevant sind, sodass MFCC in vielen Bioakustik-Aufgaben nicht zwingend besser als Log-Mel abschneiden.

**Raw Waveforms** verzichten vollständig auf eine vorgelagerte Feature-Berechnung und geben dem Modell das Audiosignal direkt im Zeitbereich, wodurch ein end-to-end Ansatz möglich wird, bei dem das Netzwerk Filter und Merkmale selbst lernt und theoretisch optimal an den Datensatz angepasst sein kann, in der Praxis aber häufig mehr Daten, sorgfältiges Tuning und stärkere Regularisierung benötigt, zudem bei langen und verrauschten Feldaufnahmen empfindlicher auf Störungen ist und rechnerisch aufwendiger werden kann.

**PCEN** baut wie Log-Mel auf einer Mel-Spektrogramm-Darstellung auf, ersetzt die reine Log-Kompression jedoch durch eine adaptive Normalisierung und Kompression pro Frequenzband, wodurch Pegelschwankungen durch Entfernung und Aufnahmegerät sowie ein gleichmäßiger Rauschteppich oft besser abgefedert werden und leise Vogelereignisse deutlicher hervortreten können, allerdings ist das Verfahren etwas komplexer und besitzt zusätzliche Parameter, sodass der Vorteil gegenüber Log-Mel vor allem dann sichtbar wird, wenn die Aufnahmebedingungen stark variieren und Hintergrundgeräusche eine große Rolle spielen.

Für die weiteren Experimente werden Log-Mel, MFCC und Raw Waveforms verwendet, weil diese drei Ansätze eine klare Abstufung von stark vorstrukturiert bis vollständig end-to-end abdecken und dadurch Unterschiede besonders deutlich werden. Log-Mel steht dabei als naheliegende Referenz, MFCC bildet als bewusst stärker verdichtete Variante derselben spektralen Ausgangsbasis einen Kontrast, bei dem sich der Effekt von Informationsreduktion direkt in den Ergebnissen und Fehlerbildern zeigt, und Raw ergänzt den Vergleich als grundlegend anderer Zugang im Zeitbereich, bei dem das Modell die Merkmalsbildung selbst übernimmt und damit sichtbar wird, wie viel der Leistung aus der gewählten Repräsentation stammt und wie viel das Netzwerk ohne explizite Feature-Berechnung kompensieren kann. PCEN wird nicht weiter betrachtet, weil es in diesem Vergleich weniger einen eigenständigen methodischen Gegenpol darstellt, sondern eine zusätzliche, parameterabhängige Ausprägung innerhalb der Mel-basierten Spektrogrammfamilie wäre, wodurch die Interpretation eher in Richtung Tuning-Effekte als in Richtung prinzipieller Repräsentationsunterschiede verschoben würde.


###3.3.1 Log-Mel-Spektogramme

Zur Berechnung der Log-Mel-Spektrogramme werden zentrale Transformationsparameter definiert.  
Die gewählten Werte stellen einen Kompromiss zwischen zeitlicher und frequenzieller Auflösung dar und orientieren sich an gängigen Konfigurationen in der Audioverarbeitung.

```python
# Transformationsparameter festlegen

n_fft = 2048        # Länge des FFT-Fensters (bestimmt Frequenzauflösung)
hop_length = 512    # Schrittweite zwischen zwei Fenstern (bestimmt Zeitauflösung)
n_mels = 128        # Anzahl der Mel-Frequenzbänder
```

Zur Wiederverwendbarkeit wird eine Funktion definiert, die ein einzelnes Segment in ein Log-Mel-Spektrogramm transformiert.

```python
def compute_log_mel(segment, sr):
    """
    Berechnet ein Log-Mel-Spektrogramm für ein gegebenes Audiosegment.
    """

    # Mel-Spektrogramm berechnen (Leistungsspektrum auf Mel-Skala)
    mel_spec = librosa.feature.melspectrogram(
        y=segment,
        sr=sr,
        n_fft=n_fft,
        hop_length=hop_length,
        n_mels=n_mels
    )

    # Logarithmische Kompression (dB-Skalierung)
    log_mel = librosa.power_to_db(mel_spec, ref=np.max)

    return log_mel
```

Anschließend wird die Transformation auf sämtliche segmentierten Waveforms angewendet.  
Das Ergebnis ist eine Sammlung zweidimensionaler Zeit-Frequenz-Repräsentationen.

```python
X_logmel = []

# Iteration über alle Segmente
for segment in X_segments:
    logmel = compute_log_mel(segment, sr)
    X_logmel.append(logmel)

# In NumPy-Array umwandeln
X_logmel = np.array(X_logmel)

# Dimension ausgeben
print("Shape der Log-Mel-Daten:", X_logmel.shape)
```

##4.1 Wahl des Modelltyps

Für die Klassifikation von Vogelstimmen wird in dieser Arbeit ein Convolutional Neural Network (CNN) als Modelltyp gewählt. Der zentrale Grund ist, dass sich Audiosignale nach einer geeigneten Vorverarbeitung als zweidimensionale Zeit-Frequenz-Repräsentationen darstellen lassen. Diese Repräsentationen ähneln Bildern, sodass CNNs ihre Stärke, das Lernen lokaler Muster wie harmonische Strukturen, Formante, zeitliche Modulationen, besonders gut ausspielen können. CNNs sind dabei robust gegenüber kleinen Verschiebungen in Zeit und Frequenz, sog. Translationsinvarianz, was bei realen Aufnahmen mit ihren verschiedenen Distanzen, Hintergrundgeräuschen und leichten Tempo-/Tonhöhenschwankungen vorteilhaft ist.

Die Wahl von CNNs wird auch durch die Literatur gestützt. *Hidayat et al. (2021)* zeigen für die Klassifikation von Eulenrufen, dass CNN-basierte Ansätze auf Spektrogramm-/MFCC-Features sehr hohe Erkennungsleistungen erreichen. Außerdem berichten sie, dass die Kombination verschiedener Merkmalsdarstellungen, z. B. Mel-Spektrogramm und MFCC, in einem Dual-Input-CNN die Performance gegenüber einzelnen Feature-Eingängen weiter verbessern kann. Dieses Ergebnis motiviert, neben einem 'klassischen' CNN auch Varianten zu berücksichtigen, die unterschiedliche Audio-Features gemeinsam auswerten.

Auch *Carvalho & Gomes (2022)* und verwandte Arbeiten ordnen CNNs als leistungsfähige Deep-Learning-Modelle für Bioakustik-Aufgaben ein, da sie aus spektralen Audioabbildungen automatisch diskriminative Merkmale lernen und dabei häufig klassische ML-Pipelines übertreffen.
